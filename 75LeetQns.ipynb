```python
class Solution:
    def climbStairs(self, n: int) -> int:
        count = 0
        if n <= 2 :
            return n
        step1, step2 = 1,2

        for _ in range (3,n+1):
            current = step1 + step2
            step1 = step2
            step2 = current

        return step2
```
