```python
# Binary Cross-Entropy Loss for scalar
def binary_cross_entropy_loss(prediction, target):
    epsilon = 1e-8  # To prevent log(0)
    prediction = torch.clamp(prediction, epsilon, 1 - epsilon)
    return -(target * torch.log(prediction) + (1 - target) * torch.log(1 - prediction))
```

```python
# Compute binary cross-entropy loss
loss = binary_cross_entropy_loss(y_pred, y)
loss
```

```python
loss.backward()
```

```python
import torch
import torch.nn.functional as F

x = torch.tensor(6.7)
y = torch.tensor(0.0)                # Actual: Not Placed

w = torch.tensor(1.0, requires_grad=True)
b = torch.tensor(0.0, requires_grad=True)

learning_rate = 0.1

print("Step | Loss    | y_pred  | w       | b")
print("-" * 50)

for step in range(1, 11):
    # 1. Forward Pass
    z = w * x + b
    y_pred = torch.sigmoid(z)
    loss = F.binary_cross_entropy(y_pred, y)

    # 2. Backward Pass (Gradients nikalna)
    loss.backward()

    # 3. Update Weights
    with torch.no_grad():
        w -= learning_rate * w.grad
        b -= learning_rate * b.grad

    # 4. Gradients ko zero karna (important!)
    w.grad.zero_()
    b.grad.zero_()

    print(f"{step:4d} | {loss.item():.4f} | {y_pred.item():.4f} | {w.item():.4f} | {b.item():.4f}")
```

```python
import torch
import torch.nn.functional as F

x = torch.tensor(6.7)
y = torch.tensor(0.0)                # Actual label

w = torch.tensor(1.0, requires_grad=True)
b = torch.tensor(0.0, requires_grad=True)

z = w * x + b
y_pred = torch.sigmoid(z)

# Loss calculate
loss = F.binary_cross_entropy(y_pred, y)

print(loss.item())     # Output: approximately 6.725
```

```python
loss.backward()     # Yeh line gradients nikal deti hai

print("w.grad =", w.grad.item())   # ≈ 6.692
print("b.grad =", b.grad.item())   # ≈ 0.9988
```

```python
learning_rate = 0.1

# Gradients already nikal chuke hain (loss.backward() ke baad)
with torch.no_grad():          # Update ke time gradient track mat karo
    w -= learning_rate * w.grad
    b -= learning_rate * b.grad

print("New w =", w.item())     # ≈ 0.3308
print("New b =", b.item())     # ≈ -0.0999
```
