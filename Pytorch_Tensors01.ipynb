Tensor Shape

```python
import torch
x = torch.tensor([[1,2,3],[4,5,6]])
x
```

```python
x.shape
```

Tensor DataType

```python
#Find dataType
x.dtype
```

```python
#assign dataType
torch.tensor([1.0,2.0,3.0],dtype=torch.int32)
torch.tensor([1,2,3],dtype=torch.float64)
```

Mathematical operations

```python
#Addition
x + 2

# division
x / 2
```

```python
a = torch.rand(2,3)
b = torch.rand(2,3)

print(a)
print(b)
```

```python
#Addition
a + b

#power
a ** b

#mod
a % b
```

```python
#sqrt
torch.sqrt(x)
```

```python
k = torch.randint(size=(2,3), low = 0, high=10, dtype=torch.float32 )
```

```python
#softmax
torch.softmax(x,dim=0)
```

```python
#relu
torch.relu(k)
```
