```python
import  numpy as np

#1D vector
s = np.array([1,2,5])
print(type(s))

#2D metrics
h = np.array([[14,8,0],[4,7,8]])
print(h)

#3D- tensor
j = np.array([[2,6,8],[4,7,9],[2,6,8]])
print(j)
```

```python
#float data type
np.array([1,4,7,8,9],dtype=float)


#boolean data type
np.array([1,4,7,8,9],dtype=bool)

np.array([1,4,7,8,9],dtype=complex)
```

```python
#np.arange
np.arange(1,9)
np.arange(1,9,2)
```

```python
# with reshape
np.arange(1,9).reshape(4,2)
np.arange(16).reshape(2,2,2,2)
```

```python
#np.ones an np.zeros
np.ones((3,4))
np.zeros((3,4))
np.random.random((3,6))
```

```python
#np.linspace - linear distance main numbers create kar ke deta hai
np.linspace(-10,10,10,dtype=int)
```

```python
#np.identity  -- diagonal axis have value others are 0
np.identity(3)
```

```python
# ndim - number of dimensions
a1 = np.arange(10)
a2 = np.arange(12, dtype=float).reshape(3,4)
a3 = np.arange(8).reshape(2,2,2)

a1.ndim
a2.ndim
a3.ndim
```

```python
#Shape - tells us the shape of array
a2.shape
print(a3.shape)
a3
```

```python
# size
print(a2.size)
a2
```

```python
# itemsize  -- the int64 takes 8 byte to store its value
a1.itemsize
print(a1.dtype)
a3.itemsize
a2.itemsize
```

```python
#astype does a type caste for all the variables
a3.astype(np.int32)
a3.dtype
```

```python
#Scalar  operations
a3**2
#a1 + 2
```
