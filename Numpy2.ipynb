# .npy ek single NumPy array ko save karta hai.

```python
import numpy as np

arr = np.array([10, 20, 30, 40])

# Save
np.save("data.npy", arr)

loaded_arr = np.load("data.npy")

print(loaded_arr)
```

# .npz — Multiple Arrays Store Karna

```python
import numpy as np

x = np.array([1, 2, 3])
y = np.array([10, 20, 30])

np.savez("dataset.npz", x=x, y=y)

load_data = np.load("dataset.npz")


print(load_data["x"])
print(load_data["y"])
```
