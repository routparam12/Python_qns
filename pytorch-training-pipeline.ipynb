### Numpy arrays to PyTorch tensors

```python
X_train_tensor = torch.from_numpy(X_train)
X_test_tensor = torch.from_numpy(X_test)
y_train_tensor = torch.from_numpy(y_train)
y_test_tensor = torch.from_numpy(y_test)
```

### Defining the model

```python
class MySimpleNN():

  def __init__(self, X):


#   Manually setting  wights & biases
    self.weights = torch.rand(X.shape[1], 1, dtype=torch.float64, requires_grad=True)
    self.bias = torch.zeros(1, dtype=torch.float64, requires_grad=True)

  def forward(self, X):
    z = torch.matmul(X, self.weights) + self.bias
    y_pred = torch.sigmoid(z)
    return y_pred

  def loss_function(self, y_pred, y):
    # Clamp predictions to avoid log(0)
    epsilon = 1e-7
    y_pred = torch.clamp(y_pred, epsilon, 1 - epsilon)

    # Calculate loss
    loss = -(y_train_tensor * torch.log(y_pred) + (1 - y_train_tensor) * torch.log(1 - y_pred)).mean()
    return loss
```

### Important Parameters

```python
learning_rate = 0.1
epochs = 25      ##Tells us how many times do we want to run?
```

### Training Pipeline

```python
# create model
model = MysimpleNN(X_train_tensor.shape[1])

# define loop
for epoch in range(epochs):

  # forward pass
  y_pred = model.forward(X_train_tensor.float()) # Ensure input is float

  # loss calculate
  loss = model.loss_function(y_pred, y_train_tensor.float()) # Ensure target is float

  # backward pass
  loss.backward()

  # parameters update
  with torch.no_grad():
    model.linear.weight -= learning_rate * model.linear.weight.grad
    model.linear.bias -= learning_rate * model.linear.bias.grad

  # zero gradients
  model.linear.weight.grad.zero_()
  model.linear.bias.grad.zero_()

  # print loss in each epoch
  print(f'Epoch: {epoch + 1}, Loss: {loss.item()}')
```

```python
import torch.nn as nn

class MysimpleNN(nn.Module):


  def __init__(self, num_features):

    super().__init__()
    self.linear = nn.Linear(num_features, 1)
    self.sigmoid = nn.Sigmoid()

  def forward(self, features):

    out = self.linear(features)
    out = self.sigmoid(out)
    return out

  def loss_function(self, y_pred, y):
  #clamp prediction to avoid log(0)
    epsilon = 1e-7
    y_pred = torch.clamp(y_pred, epsilon, 1-epsilon)
  #calculate loss
    loss = -(y * torch.log(y_pred) + (1-y) * torch.log(1-y_pred)).mean()
    return loss
```

##Using built in loss function & optimizer

```python
import torch.nn as nn

class MysimpleNN(nn.Module):


  def __init__(self, num_features):

    super().__init__()
    self.linear = nn.Linear(num_features, 1)
    self.sigmoid = nn.Sigmoid()

  def forward(self, features):

    out = self.linear(features)
    out = self.sigmoid(out)
    return out

```

```python
learning_rate=0.1
epochs = 25
```

```python
loss_function = nn.BCELoss()
```

```python
# create model
model = MysimpleNN(X_train_tensor.shape[1])

# define loop
for epoch in range(epochs):

  # forward pass
  y_pred = model.forward(X_train_tensor.float()) # Ensure input is float

  # loss calculate
  loss = loss_function(y_pred, y_train_tensor.view(-1,1).float()) # Ensure target is float

  # backward pass
  loss.backward()

  # parameters update
  with torch.no_grad():
    model.linear.weight -= learning_rate * model.linear.weight.grad
    model.linear.bias -= learning_rate * model.linear.bias.grad

  # zero gradients
  model.linear.weight.grad.zero_()
  model.linear.bias.grad.zero_()

  # print loss in each epoch
  print(f'Epoch: {epoch + 1}, Loss: {loss.item()}')
```

We will use torch.optim

torch.optim is a module in pytorch that provides  VARIETY OF OPTIMIZATON ALGORITHM USED TO UPDATE THE ARAMETERS OF YOUR MODEL DURING TRAINING

model.parameters() in pytorch retrieves an iterator overall the trainable parameters (weights & biases) in a model . These parameters are instances of torch.nn.parameters and include: weights and biases

```python
import torch.nn as nn

class MysimpleNN(nn.Module):


  def __init__(self, num_features):

    super().__init__()
    self.linear = nn.Linear(num_features, 1)
    self.sigmoid = nn.Sigmoid()

  def forward(self, features):

    out = self.linear(features)
    out = self.sigmoid(out)
    return out
```

```python
learning_rate=0.1
epochs = 25
```

```python
# create model
model = MysimpleNN(X_train_tensor.shape[1])

#optimizer
optimizer = torch.optim.SGD(model.parameters(), lr=learning_rate)

# define loop
for epoch in range(epochs):

  # forward pass
  y_pred = model.forward(X_train_tensor.float()) # Ensure input is float

  # loss calculate
  loss = loss_function(y_pred, y_train_tensor.view(-1,1).float()) # Ensure target is float

  # backward pass
  loss.backward()

#parameter update -
  optimizer.step()

#zero gradients
  optimizer.zero_grad()



  # print loss in each epoch
  print(f'Epoch: {epoch + 1}, Loss: {loss.item()}')
```
